```python
"""Companion code to https://realpython.com/simulation-with-simpy/

'Simulating Real-World Processes With SimPy'

Python version: 3.7.3
SimPy version: 3.0.11
"""
!pip install simpy 
import simpy
import random
import statistics

wait_times = []


class TechSupport(object):
    def __init__(self, env, num_engineer):
        self.env = env
        self.engineer = simpy.Resource(env, num_engineer)
        #self.manager = simpy.Resource(env, num_managers)
        #self.usher = simpy.Resource(env, num_ushers)

    def determine_support_contract(self, client):
        print ("Determining Support Contract for ", client)
        yield self.env.timeout(random.randint(1, 2))
        print ("Support Contract is found for ", client)

    def get_case_info(self, client):
        print ("Getting case info for", client)
        yield self.env.timeout(random.randint(2, 5))
        print ("Case info complete for", client)

    def work_case(self, client):
        print ("Working case for", client)
        yield self.env.timeout(random.randint(10, 200))
        print ("Case complete for", client)


    #def call_back(self, client):
     #   yield self.env.timeout(random.randint(3, 10))


def get_support(env, client, techsupport):
    # Moviegoer arrives at the theater
    arrival_time = env.now
    print ("Client {} arrived at {} minutes".format(client, arrival_time))
    with techsupport.engineer.request() as request:
        yield request
        yield env.process(techsupport.determine_support_contract(client))
        yield env.process(techsupport.get_case_info(client))
    print ("First pass complete for ", client)
    with techsupport.engineer.request() as request:
        yield request
        yield env.process(techsupport.work_case(client))
    print ("Second pass complete for ", client)
        

#    with techsupport.engineer.request() as request:
#        yield request
#        yield env.process(techsupport.get_case_info(client))

#    with techsupport.engineer.request() as request:
#        yield request
#        yield env.process(techsupport.log_case(client))

#    with techsupport.engineer.request() as request:
#        yield request
#        yield env.process(techsupport.call_back(client))

    wait_times.append(env.now - arrival_time)
    print ("client {} finished at {}".format(client, env.now))


def run_support(env, num_engineer):
    print (num_engineer)
    techsupport = TechSupport(env, num_engineer)
    #client = 1
    for client in range(3):
        env.process(get_support(env, client, techsupport))

    while True:
        yield env.timeout(5)  # Wait a bit before generating a new person

        client += 1
        env.process(get_support(env, client, techsupport))


def get_average_wait_time(wait_times):
    average_wait = statistics.mean(wait_times)
    # Pretty print the results
    minutes, frac_minutes = divmod(average_wait, 1)
    seconds = frac_minutes * 60
    return round(minutes), round(seconds)


def get_user_input():
    num_engineer = input("Input # of engineers working: ")
    num_specialist = input("Input # of specialists working: ")
    #num_servers = input("Input # of servers working: ")
    #num_ushers = input("Input # of ushers working: ")
    params = [num_engineer, num_specialist]
    if all(str(i).isdigit() for i in params):  # Check input is valid
        params = [int(x) for x in params]
    else:
        print(
            "Could not parse input. Simulation will use default values:",
            "\n1 cashier, 1 server, 1 usher.",
        )
        params = [1,1]
    return params


def main():
    # Setup
    random.seed(42)
    num_engineer, num_specialist = get_user_input()

    # Run the simulation
    env = simpy.Environment()
    env.process(run_support(env, num_engineer))
    env.run(until=480)

    # View the results
    mins, secs = get_average_wait_time(wait_times)
    print(
        "Running simulation...",
        f"\nThe average wait time is {mins} minutes and {secs} seconds.",
    )


if __name__ == "__main__":
    main()
```
